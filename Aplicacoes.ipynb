````python
from matplotlib.patches import FancyBboxPatch

areas = [
    ("FÍSICA", "Movimento Retilíneo Uniforme"),
    ("QUÍMICA", "Volume e Temperatura"),
    ("BIOLOGIA", "Crescimento de uma Planta"),
    ("MEDICINA", "Concentração a partir do Tempo"),
    ("ECONOMIA E ADMINISTRAÇÃO", "Custo, Receita e Lucro"),
    ("ENGENHARIA", "Inclinação de Rampas")
]

pos = [(0.5,4.0),(4.3,4.0),(8.1,4.0),(0.5,1.1),(4.3,1.1),(8.1,1.1)]

fig, ax = plt.subplots(figsize=(12,6))
ax.set_xlim(0,12)
ax.set_ylim(0,7)
ax.axis("off")

for (titulo, subtitulo), (x,y) in zip(areas,pos):
    caixa = FancyBboxPatch((x,y),3.2,1.7,
                           boxstyle="round,pad=0.03,rounding_size=0.08",
                           fill=False,linewidth=2)
    ax.add_patch(caixa)
    ax.text(x+1.6,y+1.05,titulo,ha="center",va="center",
            fontsize=10,weight="bold")
    ax.text(x+1.6,y+0.48,subtitulo,ha="center",va="center",fontsize=9)
plt.show()
````

## 2.1. Física — Movimento Retilíneo Uniforme

```{admonition} Orientações para os professores
:class: dropdown

Antes de mostrar a fórmula que representa essa situação, pergunte:

* Onde o ciclista está no instante t=0?
* Depois de 1s, quantos metros ele avançou?
* E depois de 2 s?
* O que varia?
* O que permanece constante?

A intenção é construir intuitivamente a ideia de **taxa de variação**.

Depois, associe:
* **a=5** à velocidade de 5 m/s e,
* **b=10** à posição inicial.
```

Entre as aplicações de **Funções de 1º Grau** na Física existe o **Movimento Retilíneo Uniforme**, onde a posição do móvel pode variar **linearmente** em função do tempo.<br>
<br>
A fórmula que traduz isso é: **S = So + vt**, onde: <br>
- **S** é a **posição final**; <br>
- **So** é a **posição inicial**;<br>
- **v** é a **velocidade do móvel** e,<br>
- **t** é o **tempo de movimento do móvel**.

### 2.1.1. Situação

```{admonition} Um ciclista em movimento
:class: tip

Um ciclista passa por um ponto localizado a **10 m da origem** e continua em linha reta com velocidade constante de **5 m/s**.

Se **t** é o **tempo**, em segundos, e **S(t)** é a **posição final**, em metros, como podemos representar esse movimento a partir de uma função?
```

````python
# Figura esquemática
fig, ax = plt.subplots(figsize=(10,3.5))
ax.set_xlim(-1,36)
ax.set_ylim(-1,5)
ax.axis("off")

ax.plot([0,35],[1,1],linewidth=4)

for x in range(0,36,5):
    ax.plot([x,x],[0.72,1.28],linewidth=1)
    ax.text(x,0.25,str(x),ha="center",fontsize=9)

ax.add_patch(Circle((10,1.65),0.45,fill=False,linewidth=2))
ax.add_patch(Circle((12,1.65),0.45,fill=False,linewidth=2))
ax.plot([10,11,12,10.7,10],[1.65,2.5,1.65,1.65,1.65],linewidth=2)
ax.add_patch(FancyArrowPatch((12.8,2.45),(17.5,2.45),
                             arrowstyle="->",mutation_scale=18))
ax.text(15.1,2.78,"5 m/s",ha="center")
ax.text(11,3.9,"posição inicial = 10 m",ha="center",weight="bold")
plt.show()
````

### 2.1.2. Animação — Acompanhando o Movimento

A partir da animação abaixo conseguimos perceber o que acontece com o ciclista a cada segundo que passa:

````python
fig, ax = plt.subplots(figsize=(9,3))
ax.set_xlim(0,45)
ax.set_ylim(-1,3)
ax.axhline(0,linewidth=1)
ax.set_yticks([])
ax.set_xlabel("Posição (m)")
ax.set_title("Movimento uniforme")

ponto, = ax.plot([],[],marker="o",markersize=12)
texto = ax.text(1.5,2.1,"")

def atualizar(frame):
    posicao = 10 + 5*frame
    ponto.set_data([posicao],[0])
    texto.set_text(f"t={frame} s  →  S(t)={posicao} m")
    return ponto,texto

anim = FuncAnimation(fig,atualizar,frames=7,interval=650,blit=False)
plt.close(fig)
HTML(anim.to_jshtml())
````

### 2.1.3. Determinação da lei da função e representação gráfica

Seguindo as informações que temos sobre o ciclista durante a animação, temos:
| $t$ (s) | $S(t)$ (m) |
|:---:|:---:|
| 0 | 10 |
| 1 | 10 + 1.5 = 15 |
| 2 | 10 + 2.5 = 20 |
| 3 | 10 + 3.5 = 25 |
| 4 | 10 + 4.5 = 30 |
| 5 | 10 + 5.5 = 35 |

* Logo, podemos perceber que o 10 é a **posição inicial** e a velocidade **5 m/s** indica a **taxa de variação** desse movimento.
  
* Assim, a fórmula que representa essa situação é **S(t) = 10 + 5t** ou **S(t) = 5t + 10**.

* No gráfico abaixo temos a representação gráfica dessa situação de movimento, observe:

````python
t = np.arange(0,7)
S = 5*t + 10

plt.figure(figsize=(8,5))
plt.plot(t,S,marker="o")
plt.axhline(0,linewidth=1)
plt.axvline(0,linewidth=1)
plt.xlabel("Tempo t (s)")
plt.ylabel("Posição S(t) (m)")
plt.title("Movimento uniforme: S(t)=5t+10")
plt.grid(True,alpha=0.3)
plt.show()
````

## 2.2. Química — Volume e Temperatura

```{admonition} Orientações para os professores
:class: dropdown

* Nesta seção, o objetivo é explorar **funções lineares**, ou seja, funções do tipo $f(x)=ax$, isto é, com b=0.

* Destaque a unidade do coeficiente angular: $0{,}08\ \text{l/K}$. Pergunte aos estudantes como ler essa unidade em palavras.

* Importante lembrar que funções de 1º grau que são **lineares** apresentam **proporcionalidade direta** em seus valores.
```

Na Química existe uma relação entre **temperatura e volume de um gás** que, sob determinadas condições, é um exemplo de aplicação de **funções de 1º grau lineares**.<br>
<br>
A fórmula que traduz isso é: **V(t) = at**, onde: <br>
- **V(t)** é o **volume do gás em litros**; <br>
- **a** é o **coeficiente de variação que indica qual é o volume de gás que se expande a cada 1 grau Kelvin de temperatura** e,<br>
- **t** é a **temperatura em graus Kelvin**.

### 2.2.1. Situação

```{admonition} Um ciclista em movimento
:class: tip

Considere um experimento em que o volume de uma amostra gasosa seja representado por **V(T)=0,08T** e onde:
* T é a temperatura em kelvin;
* V(T) é o volume em litros.

O coeficiente **0,08** indica **o coeficiente de variação do gás**. Isso significa que o gás expande de forma a aumentar **0,08 l a cada 1 grau kelvin de temperatura**.

Observe, a seguir, a tabela com valores que obedecem a essa lei:
```

| $T$ (K) | $V(T)$ (l) |
|:---:|:---:|
| 250 | 20 |
| 275 | 22 |
| 300 | 24 |
| 325 | 26 |
| 350 | 28 |

````python
T = np.array([250,275,300,325,350])
V = 0.08*T

plt.figure(figsize=(8,5))
plt.scatter(T,V,s=70)
plt.plot(T,V)
plt.xlabel("Temperatura T (K)")
plt.ylabel("Volume V(T) (l)")
plt.title("Modelo linear: volume x temperatura")
plt.grid(True,alpha=0.3)
plt.show()
````

### 2.2.2. Animação — O Volume Aumentando

Na animação abaixo podemos perceber o volume do gás aumentando à medida que a temperatura aumenta, comportamento que se explica devido à **proporcionalidade direta** existente nessa situação.

````python
fig, ax = plt.subplots(figsize=(7,5))
ax.set_xlim(0,10)
ax.set_ylim(0,10)
ax.axis("off")

ax.plot([3,3],[1,8],linewidth=2)
ax.plot([7,7],[1,8],linewidth=2)
ax.plot([3,7],[1,1],linewidth=2)

gas = Rectangle((3,1),4,3,alpha=0.22)
ax.add_patch(gas)

pistao = Rectangle((3,4),4,0.35,alpha=0.4)
ax.add_patch(pistao)

texto = ax.text(5,9,"",ha="center")

def atualizar(frame):
    temp = 250 + 10*frame
    vol = 0.08*temp
    altura = 3 + 0.06*frame
    gas.set_height(altura)
    pistao.set_y(1+altura)
    texto.set_text(f"T={temp} K  →  V={vol:.1f} L")
    return gas,pistao,texto

anim = FuncAnimation(fig,atualizar,frames=11,interval=500,blit=False)
plt.close(fig)
HTML(anim.to_jshtml())
````

```{admonition} Orientação para os professores

* Explore com seus alunos o processo de crescimento da planta: 
  * A altura inicial de 12 cm;

* Ressaltar para os alunos que modelos lineares em Biologia podem ser simplificações de situações mais complexas.

* Pergunte: **“A planta crescerá 2 cm por semana para sempre?”** A intenção é introduzir a ideia de **limite de validade da função utilizada**.
```

Entre as aplicações de **Funções de 1º Grau** na Biologia existe uma situação que pode ser modelada linearmente em um determinado intervalo de observação: a análise do **desenvolvimento de uma planta**.<br>
<br>
A fórmula que traduz isso é: **h(t) = ho + vt**, onde: <br>
- **h(t)** é a **altura final da planta depois do intervalo de tempo observado**; <br>
- **ho** é a **altura no início da observação**;<br>
- **v** é a **velocidade de crescimento da planta** e,<br>
- **t** é o **tempo de observação**.

### 2.3.1. Situação

```{admonition} Acompanhando o crescimento de uma muda de planta
:class: tip

Uma muda possui **12 cm** de altura e cresce aproximadamente **2 cm por semana** durante o período observado de **t semanas**.

Assim, a fórmula da função de 1º grau que representa essa situação é **h(t)=2t+12**, pois **a=2**, que é o **coeficiente de variação** e b=12, que é **a altura inicial** considerada.

Observe o gráfico dessa situação:
```

````python
t = np.arange(0,9)
h = 2*t + 12

plt.figure(figsize=(8,5))
plt.plot(t,h,marker="o")
plt.xlabel("Tempo t (semanas)")
plt.ylabel("Altura h(t) (cm)")
plt.title("Crescimento aproximado de uma planta")
plt.grid(True,alpha=0.3)
plt.show()
````

### 2.3.2. Animação — Crescimento no Período Observado

Na animação abaixo podemos perceber o crescimento dessa muda de planta a partir do tempo de observação em semanas:

````python
fig, ax = plt.subplots(figsize=(6,5))
ax.set_xlim(-3,3)
ax.set_ylim(0,32)
ax.set_xticks([])
ax.set_ylabel("Altura aproximada (cm)")

caule, = ax.plot([0,0],[0,12],linewidth=5)
folha1, = ax.plot([],[],linewidth=3)
folha2, = ax.plot([],[],linewidth=3)
texto = ax.text(0,29,"",ha="center")

def atualizar(frame):
    altura = 12 + 2*frame
    caule.set_data([0,0],[0,altura])
    folha1.set_data([0,1.2],[altura*0.62,altura*0.70])
    folha2.set_data([0,-1.2],[altura*0.78,altura*0.86])
    texto.set_text(f"Semana {frame}: {altura} cm")
    return caule,folha1,folha2,texto

anim = FuncAnimation(fig,atualizar,frames=9,interval=550,blit=False)
plt.close(fig)
HTML(anim.to_jshtml())
````

## 2.4. Medicina — Concentração de Uma Medicação Ao Longo do Tempo

```{admonition} Orientações para os professores
:class: dropdown

* Esta situação deve ser utilizada **somente como modelo matemático simplificado**, não como orientação clínica.

* Explore:
  * O **coeficiente angular negativo** pelo fato da concentração sempre reduzir com o passar do tempo.
  * O fato da função ser **decrescente**.
  * O **valor inicial** da função.
  * O **zero da função**: qual seu significado aqui?
 
* Pergunte: **“O que significaria uma concentração negativa? Isso faz sentido?"** Essa pergunta ajuda a discutir a necessidade de **restringir o domínio** dessa função.
```

Entre as aplicações de **Funções de 1º Grau** na **Medicina** existe uma situação que pode ser modelada linearmente em um determinado intervalo de tempo: a **concentração de um medicamento no organismo**.

A fórmula que traduz isso é: **C(t) = Co + vt**, onde:
* **C(t)** é a **concentração final do medicamento no organismo após o intervalo de tempo considerado**;
* **Co** é a **concentração inicial do medicamento no organismo**;
* **v** é a **velocidade de redução da concentração** e,
* **t** é o **intervalo de tempo considerado**.

### 2.4.1. Situação

```{admonition} Um Modelo Simplificado: Concentração de um Medicamento no Organismo
:class: tip

A concentração de um medicamento no organismo de uma pessoa reduz com o passar do tempo. Considere que a concentração de um medicamento qualquer no organismo é dado pela função **C(t)=-0,5t+10**. <br>
<br>
Assim, nesse modelo:
* A concentração inicial é 10 unidades de medicação.
* A concentração diminui 0,5 unidades por unidade de tempo.
* O coeficiente angular é negativo porque a concentração sempre diminui com o passar do tempo.<br>
<br>
No gráfico abaixo podemos acompanhar o desenvolvimento dessa situação:
```

## 2.5. Economia e Administração — Custo, Receita e Lucro

```{admonition} Orientações para os professores
:class: dropdown

* Construa as funções em três etapas:
  * **Custo:** Qual é o custo de cada unidade produzida?
  * **Receita:** Quanto é recebido por cada unidade vendida?
  * **Lucro:** Qual é a diferença entre receita e custo?

* A intenção é evitar que **L(x)=R(x)-C(x)** apareça apenas como uma regra decorada.

* Essa situação também permite conversar sobre **ponto de equilíbrio** entre funções.
```

### 2.5.1. Situação

```{admonition} Produção de peças 
:class: tip

Uma pequena empresa possui:
* **Custo fixo mensal** de R\$ 4.000,00;
* **Custo por peça produzida** de R\$ 30,00 e,
* **Preço de venda por peça** de R\$ 70,00.

Então, com base nessas informações, podemos definir três funções de 1º grau onde **x** é o número de peças produzidas:
* A função que define o dinheiro arrecadado com as vendas, que chamamos de **receita**: **R(x)=70x**.
* A função que define o dinheiro gasto para a produção dessas peças e manutenção da empresa, que chamamos de **custo**: **C(x)=30x+4000**.
* A função que define o lucro obtido pela empresa com essas **x** peças: **L(x)=R(x)-C(x)**, ou seja, **L(x)=70x-(30x+4000)=70x-30x-4000=40x-4000**.

No gráfico abaixo vamos representar as funções R(x) e C(x). Analisando o gráfico podemos perceber que a intersecção das duas retas é o **ponto de equilíbrio**, ou seja, **para qual número de peças a receita é igual aos custos**.
```

````python
x = np.linspace(0,160,161)
R = 70*x
C = 20*x + 4000

plt.figure(figsize=(9,5))
plt.plot(x,R,label="Receita R(x)=70x")
plt.plot(x,C,label="Custo C(x)=30x+4000")
plt.xlabel("Quantidade de peças")
plt.ylabel("Valor (R$)")
plt.title("Receita e custo")
plt.grid(True,alpha=0.3)
plt.legend()
plt.show()
````

````python
from IPython.display import display, HTML

texto = "Fonte: Jupyter Notebook"
# Alinha à direita, define a fonte como 11px e muda a cor para cinza
display(HTML(f'<p style="text-align: right; font-size: 11px; color: gray;">{texto}</p>'))
````

* Quando **receita=custos** temos o **ponto de equilíbrio** da situação:<br>
**70x=30x+4000**  
**40x=4000**    
**x=100** <br>
* Então, com **100 peças**, o lucro é igual a **zero**. <br> 

* A partir disso também podemos definir que:
  * Com um **número de peças < 100**: a receita **é menor** que os custos, logo a empresa terá **prejuízo**.
  * Com um **número de peças > 100**: a receita **é maior** que os custos, logo a empresa terá **lucro**.<br>
<br>
* Observe todas essas características no gráfico abaixo:

````python
x = np.linspace(0,160,161)
L = 40*x - 4000

plt.figure(figsize=(9,5))
plt.plot(x,L,label="Lucro L(x)=40x-4000")
plt.axhline(0,linewidth=1)
plt.axvline(0,linewidth=1)
plt.scatter([100],[0],s=120,label="Ponto de equilíbrio")
plt.xlabel("Quantidade de peças")
plt.ylabel("Lucro (R$)")
plt.title("Função lucro")
plt.grid(True,alpha=0.3)
plt.legend()
plt.show()
````

### 2.5.2. Animação — Passando do Prejuízo Para o Lucro

````python
fig, ax = plt.subplots(figsize=(9,5))
ax.set_xlim(0,150)
ax.set_ylim(-4500,4000)
ax.plot([0,150],[-4000,3500],linewidth=2)
ax.axhline(0,linewidth=1)
ax.axvline(0,linewidth=1)
ax.grid(True,alpha=0.3)
ax.set_xlabel("Quantidade de peças")
ax.set_ylabel("Lucro (R$)")

ponto, = ax.plot([],[],marker="o",markersize=10)
texto = ax.text(5,3200,"")

def atualizar(frame):
    q = frame*5
    lucro = 50*q - 4000
    status = "prejuízo" if lucro < 0 else ("equilíbrio" if lucro == 0 else "lucro")
    ponto.set_data([q],[lucro])
    texto.set_text(f"x={q}  →  L(x)=R$ {lucro:.0f}  →  {status}")
    return ponto,texto

anim = FuncAnimation(fig,atualizar,frames=31,interval=300,blit=False)
plt.close(fig)
HTML(anim.to_jshtml())
````

## 2.6. Engenharia — Inclinação de Uma Rampa

```{admonition} Orientações para os professores
:class: dropdown

* Relacione o coeficiente angular à razão:
\begin{equation*}
a=\frac{\Delta y}{\Delta x}.
\end{equation*}

* Se a distância horizontal aumenta 5 m, a altura aumenta:
\begin{equation*}
0{,}20\cdot5=1\text{ m}.
\end{equation*}

* Logo:
\begin{equation*}
a=\frac{1}{5}=0{,}20.
\end{equation*}

* A intenção é conectar **Álgebra, Geometria e interpretação gráfica**.
```

Entre as aplicações de **Funções de 1º Grau** na **Engenharia** existe uma situação bastante utilizada em nosso cotidiano: a **inclinação de rampas**.

A fórmula que traduz isso é: **h(x) = ax**, onde:
* **h(x)** é a **altura da rampa**;
* **a** é o **coeficiente de inclinação da rampa**;
* **x** é o **deslocamento horizontal considerado**.

### 2.6.1. Situação

```{admonition} Um Modelo de Rampa
:class: tip

Considere uma rampa cuja altura aumenta **0,20 m** para cada **1 m** de deslocamento horizontal.

Nessa situação, a altura pode ser representada por: **h(x)=0,20x**.

A representação gráfica dessa situação é:
```

````python
fig, ax = plt.subplots(figsize=(9,5))
ax.set_xlim(-0.5,10.5)
ax.set_ylim(-0.5,3.2)

x = np.linspace(0,10,100)
h = 0.20*x

ax.plot(x,h,linewidth=3)
ax.plot([0,5,5],[0,0,1],linestyle="--")
ax.text(2.5,-0.25,"Δx = 5 m",ha="center")
ax.text(5.25,0.5,"Δh = 1 m",va="center")

ax.axhline(0,linewidth=1)
ax.axvline(0,linewidth=1)
ax.set_xlabel("Distância horizontal x (m)")
ax.set_ylabel("Altura h(x) (m)")
ax.set_title("Rampa modelada por h(x)=0,20x")
ax.grid(True,alpha=0.3)
plt.show()
````
