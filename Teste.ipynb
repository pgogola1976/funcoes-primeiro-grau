* Nessa simples brincadeira com palitos de fósforo podemos observar o que a Matemática chama de **relação de dependência entre grandezas** porque **o número de palitos de fósforo que formam a figura DEPENDE da posição da figura na sequência**, como vemos na tabela abaixo:

| Posição da figura(n) | Número de palitos de fósforo utilizados(p) |
| :---: | :---: |
| 1 | 4 |
| 2 | 7 |
| 3 | 10 |
| 4 | 13 |

    <ul style="margin: 0; padding-left: 20px; color: #333;">
        <li>A <b>variável independente</b> é representada pela <b>posição da figura na sequência(n)</b> e a <b>variável dependente</b> é representada pelo <b>número de palitos de fósforo utilizados para formar a figura(p)</b>.</li>

````python
from jupyterquiz import display_quiz

pergunta1 = [{
        "question": "Observando essa tabela e chamando a posição da figura de n e o número de palitos utilizados de p, você consegue encontrar a fórmula que determina os valores de p em função de n?",
        "type": "multiple_choice",
        "answers": [
            {
                "answer": "p = 2n + 2",
                "correct": False,
                "feedback": "Tente novamente, essa não é a resposta correta!"
            },
            {
                "answer": "p = 4n",
                "correct": False,
                "feedback": "Tente novamente, essa não é a resposta correta!"
            },
            {
                "answer": "p = 5n - 1",
                "correct": False,
                "feedback": "Tente novamente, essa não é a resposta correta!"
            },
            {
                "answer": "p = 3n + 1",
                "correct": True,
                "feedback": "Você acertou! Essa é a resposta correta!"
            }
                ]
}]

display_quiz(pergunta1)
````

```{admonition} Para entender a resposta certa, clique nessa aba!
:class: dropdown

<div class="alert alert-block alert-info" style="text-align: justify;">

* Observe o número de palitos utilizados na construção das figuras dessa sequência: **a cada nova figura aumentamos 3 palitos**, ou seja, a **variável dependente** apresenta uma **variação constante** igual a 3.  

* Essa **variação constante** igual a 3 nos diz que o número de palitos de fósforo utilizados estão ligados aos **múltiplos de 3**, ou seja, **3n**. **Essa variação constante sempre fica ligada à variável independente.** Porém, nessa situação, para cada valor de posição **n**, além de fazermos **3n**, deveremos fazer um pequeno ajuste e **acrescentar 1** para encontrarmos os valores representados na tabela. Assim sendo, a **regra de formação** que representa essa relação de dependência é: **p = 3n + 1**.

| Posição da figura(n) | Cálculos envolvidos | Número de palitos de fósforo utilizados(p) |
| :---: | :---: | :---: |
| 1 | 3 . 1 + 1 | 4 |
| 2 | 3 . 2 + 1 | 7 |
| 3 | 3 . 3 + 1 | 10 |
| 4 | 3 . 4 + 1 | 13 |

</div>
```

## 2.2. Definição

```{admonition} Função polinomial de 1º grau ou função afim
:class: hint
Uma **função polinomial de 1º grau**, ou **função afim**, é uma função f de R em R dada por uma lei de forma **f(x) = ax + b**, em que **a** e **b** são números reais e **a** é não-nulo.

Os números **a** e **b** são chamados de **coeficientes**: 
* **a** é chamado de **coeficiente angular** ou **coeficiente de variação**,
* **b** é chamado de **coeficiente linear**.

Quando **b=0** temos a função de 1º grau em seu formato **f(x)=ax**, o que garante uma característica a mais para esse tipo de função: **proporcionalidade direta**. Dizemos, então, que essa função é uma **função linear**. No gráfico, a característica principal das retas que representam funções lineares é o fato de que elas sempre passam pelo ponto (0,0).
```
