```{admonition} Corrida de táxi
Chamando de x a distância percorrida, em quilômetros, e de P(x) o preço total da corrida, em reais, temos

                                        P(x) = 2,50x + 6

Antes de pensar no gráfico, vamos organizar alguns valores em uma tabela:
```

```{admonition} Clique para entender!
:class: dropdown

<div class="alert alert-block alert-info" style="text-align: justify;">

A cada aumento de 1 km na distância, o preço aumenta R$ 2,50.

Esse valor é a **taxa de variação constante** da função. Na expressão $P(x)=2,50x+6$, ele aparece multiplicando a variável $x$.

Por outro lado, quando $x=0$, temos $P(0)=6$. Isso significa que a corrida já começa com um valor inicial de R$ 6,00.

Portanto:

* $2,50$ indica quanto o preço aumenta a cada quilômetro,
* $6,00$ indica a taxa fixa (ou bandeirada) da corrida.
```

### 1.2. Do quadro de valores ao gráfico

```{admonition} Marcando pontos no plano cartesiano
:class: tip
<div class="alert alert-block alert-success" style="text-align: justify;">

Cada linha da tabela gera um par ordenado (x,y), onde y=P(x).

Por exemplo, se x=2, então P(2)= 2,50.2 + 6 = 5 + 6 = 11. Logo, temos o ponto (2,11).

Pela tabela acima, já temos 5 pontos dessa função: (0;6), (1;8,50), (2;11), (3;13,50), (4;16) e (5;18,50).

Marcando esses pontos no plano cartesiano podemos visualizar uma característica muito importante da função de 1º grau. O que você consegue perceber?

</div>
```

### 1.3. O gráfico de uma função de 1º grau

```{admonition} Função de 1º grau e seu gráfico
:class: hint
Uma **função de 1º grau** é uma função dada por uma lei da forma

$f(x)=ax+b,$

em que **a** e **b** são números reais e **$a\neq 0$**. <br>

Esses valores de **a** e **b** são chamados de **coeficientes**. <br>

O gráfico de uma função de 1º grau é uma **reta**. <br>

Exemplos:
* Na função f(x) = 5x + 7, **a = 5** e **b = 7**.
* Na função f(x) = -3x + 1, **a = - 3** e **b = 1**.
* Na função f(x) = 4 + 2x, **a = 2** e **b = 4**.

```

```{admonition} Como desenhar o gráfico de uma função de 1º grau?
:class: tip
<div class="alert alert-block alert-success" style="text-align: justify;">

Como o gráfico é uma **reta**, uma primeira forma mais simples de desenhar o gráfico é encontrar dois pontos da função e traçar a reta que passa por eles.

Para isso podemos escolher dois valores de $x$, calcular os valores correspondentes de $f(x)$ e marcar os dois pontos no plano cartesiano.

</div>
```

### 1.4. Exemplo

Vamos construir o gráfico da função $f(x)=2x+1$.

Escolhendo $x=0$ e $x=2$, temos:

| $x$ | $f(x) = 2.x + 1$ | Ponto |
| :---: | :---: | :---: |
| 0 | $f(0) = 2.0 + 1 = 1$ | $(0,1)$ |
| 2 | $f(2) = 2.2 + 1 = 5$ | $(2,5)$ |

Assim, o gráfico é a reta que passa pelos pontos $(0,1)$ e $(2,5)$.

````python
import matplotlib.pyplot as plt

x = [-1, 0, 1, 2, 3]
y = [2*i + 1 for i in x]

plt.figure(figsize=(7,4))
plt.plot(x, y, marker='o')
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Gráfico de f(x)=2x+1')
plt.grid(True)
plt.show()
````

## 2. Coeficientes de uma função de 1º grau

### 2.1. Coeficiente a: Coeficiente angular

```{admonition} Coeficiente a: Inclinação da reta ou taxa de variação
:class: hint
Na função $f(x)=ax+b$, o número $a$ é chamado de **coeficiente angular** ou **coeficiente de variação**.

Além de indicar a **taxa de variação** da função, ele determina a **inclinação da reta**. 

* Se $a>0$, a função é **crescente** e a reta é **inclinada para a direita**.
* Se $a<0$, a função é **decrescente** e a reta é **inclinada para a esquerda**.

* A função é dita **crescente** porque quando **x aumenta**, o valor de **y também aumenta**.
* A função é dita **decrescente** porque quando **x aumenta**, o valor de **y diminui**.
```

```{admonition} Subindo ou descendo?
:class: tip
<div class="alert alert-block alert-success" style="text-align: justify;">

Considere a função $f(x)=ax+b$.

O número **a** indica **quanto o valor de f(x) varia quando o valor de x aumenta uma unidade**, por isso esse número é conhecido como **coeficiente de variação**.

**Exemplo:**
* Na função f(x) = 2x + 3, **o coeficiente angular é 2**, ou seja, **a=2**. Isso quer dizer que cada vez que **o valor de x aumenta 1 unidade**, **o valor de y ou f(x) aumenta 2 unidades**. Na tabela a seguir podemos ver isso, enquanto x varia de 1 em 1, y varia de 2 em 2:
```

### 2.2. Coeficiente b: Coeficiente linear

```{admonition} Coeficiente linear: onde a reta cruza o eixo vertical
:class: hint
Na função $f(x) = ax + b$, o número **b** é chamado de **coeficiente linear**.

Ele indica o ponto em que a reta **cruza o eixo y**, isto é, o ponto $(0,b)$.
```

```{admonition} Olhando para o ponto inicial
:class: tip
<div class="alert alert-block alert-success" style="text-align: justify;">

Considere a função $f(x)=ax+b$.

Quando **x=0**, temos **f(0) = a . 0 + b = b**.

Isso significa que o número **b** é o valor da função quando **x=0**.

No gráfico, esse valor indica onde a reta **intercepta o eixo vertical**, ou seja, o **eixo y**.

Em situações do cotidiano ou das Ciências em geral, o valor de b também pode ser **o "valor inicial"** daquela situação.

</div>
```

#### 3.3.2. Atividade 4: Comparando retas com o mesmo coeficiente linear

* No gráfico abaixo percebemos três funções representadas: f(x) = x + 1, g(x) = 2x + 1 e h(x) = - x + 1.

* Em todas as funções representadas, os valores dos coeficientes lineares é o mesmo, **b = 1**, embora os coeficientes angulares sejam diferentes. O que isso acarretou aos gráficos?

* Essas funções são crescentes ou decrescentes? Por quê?


````python
import matplotlib.pyplot as plt

x = [-3, -2, -1, 0, 1, 2, 3]
funcoes = {
    'f(x)=x+1': [i + 1 for i in x],
    'g(x)=2x+1': [2*i + 1 for i in x],
    'h(x)=-x+1': [-i + 1 for i in x]
}

plt.figure(figsize=(7,4))
for nome, valores in funcoes.items():
    plt.plot(x, valores, marker='o', label=nome)
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Comparando retas com o mesmo coeficiente linear')
plt.grid(True)
plt.legend()
plt.show()
````

```{admonition} Vamos ver se você acertou a Atividade 4?
* Como elas possuem **coeficientes angulares diferentes**, temos:

    * Em $f(x)=x+1$, o coeficiente angular é $1$, logo a reta é **inclinada para a direita** e a função é **crescente**.
    
    * em $g(x)=2x+1$, o coeficiente angular é $2$, logo a reta é **inclinada para a direita** e a função é **crescente**.
    
    * em $h(x)=-x+1$, o coeficiente angular é $-1$, logo a reta é **inclinada para a esquerda** e a função é **decrescente**.
```

## 5. Resumo da Seção

```{admonition} O que aprendemos?
:class: hint
Nesta seção, vimos que:

* O gráfico de uma função de 1º grau é uma **reta**.

* O coeficiente linear $b$ indica onde a reta **cruza o eixo y**.

* O coeficiente angular $a$ indica a **taxa de variação** e a **inclinação da reta**.

* Se **a>0**, a função é **crescente** e a reta fica inclinada para a **direita**.

* Se **a<0**, a função é **decrescente** e a reta fica inclinada para a **esquerda**.

* Dois pontos são suficientes para desenhar o gráfico de uma função de 1º grau.
```
